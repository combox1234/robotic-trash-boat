## Train YOLOv8

```python
from ultralytics import YOLO\n
\n
model = YOLO('yolov8n.pt')\n
results = model.train(\n
    data='/content/most_recent_vrs/data_colab.yaml',\n
    epochs=50,\n
    imgsz=640,\n
    batch=16,\n
    project='/content/most_recent_vrs/runs',\n
    name='detect_colab'\n
)
```

## Validate

```python
from ultralytics import YOLO\n
best = YOLO('/content/most_recent_vrs/runs/detect_colab/weights/best.pt')\n
metrics = best.val(data='/content/most_recent_vrs/data_colab.yaml')\n
print(metrics)
```

## Predict on sample image

```python
from ultralytics import YOLO\n
model = YOLO('/content/most_recent_vrs/runs/detect_colab/weights/best.pt')\n
res = model.predict('/content/most_recent_vrs/pb.jpeg', conf=0.25, save=True)\n
print('Saved in:', res[0].save_dir)
```
